# Sales Data Analysis

## Project Overview

This project analyzes sales data using Python and Pandas.

The objective is to understand sales performance,
identify top-performing products, and discover
monthly revenue trends.

```python
import pandas as pd

print("Libraries imported successfully!")
```

```python
import pandas as pd

# 1. Create a tiny shop's sales data (normally this comes from a CSV)
df = pd.DataFrame({
    "date":     ["2026-01-05", "2026-01-12", "2026-02-03", "2026-02-18", "2026-02-25"],
    "product":  ["Laptop", "Mouse", "Laptop", "Keyboard", "Mouse"],
    "price":    [900, 20, 900, 50, 20],
    "quantity": [2, 10, 1, 4, 5],
})

# 2. Look at it: always check what you have first
print(df.head())

# 3. Clean: turn the date text into real dates so we can group by month
df["date"] = pd.to_datetime(df["date"])

# 4. New column: money earned per sale
df["revenue"] = df["price"] * df["quantity"]

# 5a. Question: which product earns the most?
print(df.groupby("product")["revenue"].sum().sort_values(ascending=False))

# 5b. Question: is revenue going up or down each month?
df["month"] = df["date"].dt.to_period("M")
print(df.groupby("month")["revenue"].sum())
```
